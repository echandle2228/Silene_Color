```python
import yaml

# Retrieve yaml from roboflow
dataset_root = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/Annotated_WhiteChips/ColorStandardFinder-2'
yaml_path = f'{dataset_root}/data.yaml'

data_config = {
    'train': f'{dataset_root}/train/images',
    'val': f'{dataset_root}/train/images',
    'nc': 1,
    'names': ['white_chip']
}

with open(yaml_path, 'w') as f:
    yaml.dump(data_config, f, default_flow_style=False)
```

```python
from ultralytics import YOLO

#Load the model
model = YOLO('yolov8n.pt')

#Train the
model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    device=0,        # This ensures it uses the T4 GPU
    project='/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/Roboflow_Dataset/ColorStandardFinder-2',
    name='white_chip_detector'
)
```

```python
import os

# The top-level folder where you saved the V2 training results
SEARCH_PATH = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/Roboflow_Dataset'

print("🔍 Searching for your V2 weights...")
found_any = False

for root, dirs, files in os.walk(SEARCH_PATH):
    for file in files:
        if file == 'best.pt':
            print(f"✅ FOUND IT: {os.path.join(root, file)}")
            found_any = True

if not found_any:
    print("❌ Still nothing. Check if the training cell actually finished without errors!")
```

```python
import os
import pandas as pd
from ultralytics import YOLO

# 1. THE EXACT PATH FOUND BY YOUR SEARCHER [cite: 2026-02-26]
MODEL_PATH = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/Roboflow_Dataset/ColorStandardFinder-2/white_chip_detector2/weights/best.pt'
IMAGE_FOLDER = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/2nd2024JPG'

# 2. LOAD THE V2 BRAIN
model = YOLO(MODEL_PATH)

# 3. RUN INFERENCE (Finding chips in all 348 images)
# Setting conf=0.50 to ensure we only get high-quality detections [cite: 2026-02-04]
results = model.predict(source=IMAGE_FOLDER, conf=0.50, save=False, stream=True)

all_coords = []

print("🚀 Running V2 White Chip Inference...")
for result in results:
    filename = os.path.basename(result.path)
    boxes = result.boxes.xyxy.cpu().numpy()
    scores = result.boxes.conf.cpu().numpy()

    if len(boxes) > 0:
        # Keep only the single most confident chip per image [cite: 2026-02-04]
        best_idx = scores.argmax()
        box = boxes[best_idx]

        all_coords.append({
            'filename': filename,
            'xmin': box[0], 'ymin': box[1], 'xmax': box[2], 'ymax': box[3],
            'confidence': scores[best_idx]
        })

# 4. SAVE THE COORDINATES PACKAGE [cite: 2026-02-04]
df = pd.DataFrame(all_coords)
OUTPUT_PATH = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/YOLO_Output/white_chip_coords_v2.csv'
os.makedirs(os.path.dirname(OUTPUT_PATH), exist_ok=True)
df.to_csv(OUTPUT_PATH, index=False)

print(f"📦 SUCCESS: Detected chips in {len(df)} images. CSV saved to YOLO_Output!")
```

```python
#Save YOLO Output
PACKAGE_DIR = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/YOLO_Output'
os.makedirs(PACKAGE_DIR, exist_ok=True)

#Save to CSV
df = pd.DataFrame(all_coords)
package_path = os.path.join(PACKAGE_DIR, 'white_chip_coords.csv')
df.to_csv(package_path, index=False)

print(f"Yolo output saved properly {package_path}")
print("Head back to the Main Line (CPU) notebook!")
```
