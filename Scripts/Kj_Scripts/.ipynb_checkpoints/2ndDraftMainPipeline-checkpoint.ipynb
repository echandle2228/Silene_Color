Visualize Labeled Images

```python
import os
import random
import cv2
import matplotlib.pyplot as plt

#Point to the train folder that just downloaded with images and labels folders. Locate with dataset_location from last cell
train_images_path = os.path.join(dataset_location, 'train/images')
train_labels_path = os.path.join(dataset_location, 'train/labels')

#Get 10 Random images
image_files = [f for f in os.listdir(train_images_path) if f.endswith(('.jpg', '.jpeg'))]
random_samples = random.sample(image_files, 10)

#Set up plot (5X2)
plt.figure(figsize=(15, 25))

for i, filename in enumerate(random_samples):
    #Load the image and swap BGR to RGB (openCV shennanigans)
    img_path = os.path.join(train_images_path, filename)
    image = cv2.imread(img_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB) # BGR to RGB

    #Load the corresponding label file
    base_name = os.path.splitext(filename)[0]
    label_path = os.path.join(train_labels_path, base_name + '.txt')

    #Convert label data into coords
    if os.path.exists(label_path):
        with open(label_path, 'r') as f:
            for line in f:
                parts = line.split()
                if len(parts) > 5:
                    coords = list(map(float, parts[1:]))
                    all_x, all_y = coords[0::2], coords[1::2]
                    x1_n, y1_n, x2_n, y2_n = min(all_x), min(all_y), max(all_x), max(all_y)
                else:
                    x_c, y_c, w, h = map(float, parts[1:])
                    x1_n, y1_n, x2_n, y2_n = x_c-w/2, y_c-h/2, x_c+w/2, y_c+h/2

                #Convert to pixels
                h_img, w_img, _ = image.shape
                cv2.rectangle(image, (int(x1_n*w_img), int(y1_n*h_img)),
                              (int(x2_n*w_img), int(y2_n*h_img)), (0, 255, 0), 4)

    #Truncate filename
    clean_title = filename.split('_jpg')[0] + '.jpg'

    #Put images in the grid
    plt.subplot(5, 2, i + 1)
    plt.imshow(image)
    plt.title(f"Label Check: {clean_title}", fontsize=12)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import os
import pandas as pd
import rawpy
import numpy as np
import cv2

# --- 1. SETTINGS ---
INPUT_FOLDER = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/2nd2024RAW'
OUTPUT_FOLDER = '/content/3_Scientific_Corrected'
CSV_PATH = '/content/drive/MyDrive/1 Projects/Flowers Lab/2nd Draft Files/YOLO_Output/white_chip_coords_v2.csv' # <-- Ensure this matches your uploaded CSV
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

df_coords = pd.read_csv(CSV_PATH)

print(f"🎨 Starting color correction on {len(df_coords)} images...")

for index, row in df_coords.iterrows():
    # 1. Get the base name (e.g., 'DSC_2755') by stripping whatever extension is in the CSV
    original_name = row['filename']
    base_name = os.path.splitext(original_name)[0]

    # 2. Try to find the .NEF or .nef version in your raw folder
    raw_filename = None
    for ext in ['.NEF', '.nef']:
        potential_path = os.path.join(INPUT_FOLDER, base_name + ext)
        if os.path.exists(potential_path):
            raw_filename = base_name + ext
            file_path = potential_path
            break

    if not raw_filename:
        # If we still can't find it, print a helpful message
        print(f"⚠️ Skipping {base_name}: No .NEF found in {INPUT_FOLDER}")
        continue

    try:
        # A. DEVELOP RAW DATA
        with rawpy.imread(file_path) as raw:
            img_rgb = raw.postprocess(use_camera_wb=False, no_auto_bright=True, bright=1.0)

        # B. GET COORDS
        x1, y1 = int(row['xmin']), int(row['ymin'])
        x2, y2 = int(row['xmax']), int(row['ymax'])

        # C. THE "SAFE ZONE" SHRINK (Avoids the black plastic)
        w, h = x2 - x1, y2 - y1
        s_x1, s_x2 = int(x1 + (w * 0.3)), int(x2 - (w * 0.3))
        s_y1, s_y2 = int(y1 + (h * 0.3)), int(y2 - (h * 0.3))

        white_patch = img_rgb[s_y1:s_y2, s_x1:s_x2]

        # D. CALCULATE SCALING
        avg_rgb = np.mean(white_patch, axis=(0, 1))
        factors = avg_rgb[1] / avg_rgb

        # E. APPLY TO WHOLE IMAGE
        corrected_img = img_rgb.astype(np.float32)
        for i in range(3):
            corrected_img[:,:,i] *= factors[i]

        final_img = np.clip(corrected_img, 0, 255).astype(np.uint8)

        # F. SAVE AS PNG (Higher quality than JPG for your PhD)
        save_path = os.path.join(OUTPUT_FOLDER, f"corrected_{base_name}.png")
        cv2.imwrite(save_path, cv2.cvtColor(final_img, cv2.COLOR_RGB2BGR))

        print(f"✅ Corrected: {raw_filename}")

    except Exception as e:
        print(f"❌ Error on {raw_filename}: {e}")

print(f"\n🚀 Done! PhD-ready images are in: {OUTPUT_FOLDER}")
```

```python
import numpy as np

qc_data = []

for index, row in df_coords.iterrows():
    # Load corrected image
    corr_path = os.path.join(CORRECTED_FOLDER, f"corrected_{row['filename']}")
    img = cv2.imread(corr_path)

    if img is not None:
        # Sample the chip again
        x1, y1, x2, y2 = map(int, [row['xmin'], row['ymin'], row['xmax'], row['ymax']])
        chip = img[y1:y2, x1:x2]

        # Calculate mean R, G, B
        avg_bgr = np.mean(chip, axis=(0, 1))

        # Reliability Metric: Standard Deviation between channels
        # Lower is better!
        balance_error = np.std(avg_bgr)

        qc_data.append({
            'filename': row['filename'],
            'balance_error': balance_error,
            'avg_brightness': np.mean(avg_bgr)
        })

qc_df = pd.DataFrame(qc_data)
qc_df.to_csv(os.path.join(BASE_PATH, 'white_balance_reliability_report.csv'), index=False)

print(f"📊 Reliability Report Generated! Average Error: {qc_df['balance_error'].mean():.4f}")
```

```python
def high_precision_white_balance(image, coords):
    # 1. LOCAL SAMPLING (Avoid the edges of the chip)
    x1, y1, x2, y2 = map(int, [coords['xmin'], coords['ymin'], coords['xmax'], coords['ymax']])

    # Shrink the box by 25% on all sides to get the 'pure' center pixels
    w, h = x2 - x1, y2 - y1
    cx1, cy1 = x1 + int(w*0.25), y1 + int(h*0.25)
    cx2, cy2 = x2 - int(w*0.25), y2 - int(h*0.25)

    white_patch = image[cy1:cy2, cx1:cx2]

    # 2. GAMMA-AWARE MATH
    # Convert to float and 'undo' typical sRGB gamma (approx 2.2)
    patch_linear = np.power(white_patch / 255.0, 2.2)
    avg_rgb_linear = np.mean(patch_linear, axis=(0, 1))

    # Calculate factors in linear space
    target = np.max(avg_rgb_linear)
    factors = target / (avg_rgb_linear + 1e-6)

    # Apply to whole image in linear space
    img_linear = np.power(image / 255.0, 2.2)
    for i in range(3):
        img_linear[:,:,i] *= factors[i]

    # Re-apply gamma and clip
    corrected = np.power(np.clip(img_linear, 0, 1), 1/2.2)
    return (corrected * 255).astype(np.uint8)
```

```python
import numpy as np
import os
import cv2
import pandas as pd

# 1. SETUP PATHS
CORRECTED_V3_FOLDER = os.path.join(BASE_PATH, '4_HighPrecision_Corrected')
os.makedirs(CORRECTED_V3_FOLDER, exist_ok=True)

def high_precision_white_balance(image, coords):
    """
    Applies Gamma-Aware correction and local inner-chip sampling
    to reduce edge noise and lighting gradients.
    """
    # LOCAL SAMPLING: Shrink the detection box to avoid black plastic edges
    x1, y1, x2, y2 = map(int, [coords['xmin'], coords['ymin'], coords['xmax'], coords['ymax']])
    w, h = x2 - x1, y2 - y1
    cx1, cy1 = x1 + int(w*0.25), y1 + int(h*0.25)
    cx2, cy2 = x2 - int(w*0.25), y2 - int(h*0.25)

    white_patch = image[cy1:cy2, cx1:cx2]

    # GAMMA-AWARE MATH: Convert to linear space (undoing the sRGB curve)
    patch_linear = np.power(white_patch / 255.0, 2.2)
    avg_rgb_linear = np.mean(patch_linear, axis=(0, 1))

    # Calculate scaling factors to make the chip neutral in linear space
    target = np.max(avg_rgb_linear)
    factors = target / (avg_rgb_linear + 1e-6)

    # Apply to the WHOLE image in linear space
    img_linear = np.power(image / 255.0, 2.2)
    for i in range(3):
        img_linear[:,:,i] *= factors[i]

    # Re-apply gamma curve and clip to valid 0-255 range
    corrected = np.power(np.clip(img_linear, 0, 1), 1/2.2)
    return (corrected * 255).astype(np.uint8)

# 2. RUN BATCH PROCESS
v3_qc_data = []
print("🎨 Running High-Precision (V3) Correction on 340 images...")

for index, row in df_coords.iterrows():
    img_path = os.path.join(IMAGE_FOLDER, row['filename'])
    img = cv2.imread(img_path)

    if img is not None:
        corrected = high_precision_white_balance(img, row)

        # Save V3 Image
        out_name = f"v3_corr_{row['filename']}"
        cv2.imwrite(os.path.join(CORRECTED_V3_FOLDER, out_name), corrected)

        # Immediate QC Check for this version
        # Sample the chip center again to measure new error
        x1, y1, x2, y2 = map(int, [row['xmin'], row['ymin'], row['xmax'], row['ymax']])
        w, h = x2 - x1, y2 - y1
        cx1, cy1, cx2, cy2 = x1+int(w*.25), y1+int(h*.25), x2-int(w*.25), y2-int(h*.25)
        new_chip = corrected[cy1:cy2, cx1:cx2]
        new_avg_bgr = np.mean(new_chip, axis=(0,1))

        v3_qc_data.append({
            'filename': row['filename'],
            'v3_error': np.std(new_avg_bgr)
        })

# 3. REPORT SUCCESS
v3_qc_df = pd.DataFrame(v3_qc_data)
v3_avg_error = v3
```
